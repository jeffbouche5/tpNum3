## Question 2: Loi exponentielle de paramètre 1/2.

L'espérance 2, cf <code>exponential</code>. Pour $x > 0$, $f(x) = (1/2)e^{-x/2}$ et $F(x) = 1 - e^{-x/2}$. $F^{-1}(u) = -2 \ln(1 - u)$ où $1 - U \sim U$.

```python
N=10**4; U=rand(N,1);
X=-2*log(U);                                          ## application terme à terme du log
hist(X, bins = linspace(0,10,51), density = True); 
xx=linspace(0,10,201);
plot(xx,exp(-xx/2)/2 ,'r')
```

 Soit $U$ une variable aléatoire uniforme sur $[0,1]$. La variable aléatoire $2U$ suit donc une loi uniforme sur $[0,2]$. De façon analogue, trouver des expressions ne dépendant que de $U$ pour les lois suivantes. A chaque fois, simuler $10000$ variables aléatoires indépendantes ayant cette loi. et suivant des lois de probabilité suivantes (illustrer par un histogramme) : <br>
1) La loi uniforme sur $[-1,1]$. <br>
2) La loi exponentielle de paramètre $3$. <br>
3) La loi de Bernoulli de paramètre $1/3$. <br>
4) La loi sur $\{1,2,3,4\}$, donnant les probabilités respectives $0.1, 0.3, 0.5, 0.1$ à $1, 2, 3$ et~$4$. <br>
5) La loi sur $\{-1,3,8\}$ donnant les probabilités $0.7, 0.2, 0.1$ à $-1, 3$ et $8$. <br>
6) La loi de densité
$$f(x) = 2xe^{-x^2} \, \mathbb{1}_{[0,+\infty[}(x)$$

```python
# Correction de la question 4)
X=[1,2,3,4]
p=[0.1, 0.3, 0.5,0.1]
s=np.cumsum(p)

n= 10000 # taille de l'échantillon
Y=[]
for i in range(n):
    U=np.random.uniform()
    i=0
    while U > s[i]:
        i=i+1
    Y.append(X[i])

# Vérification
hist(Y, density=True, bins=[0.5,1.5,2.5,3.5,4.5]);
plot(X, p, 'ro' )

```

## Question 6: Box Muller. 

On simule un vecteur aléatoire $(X, Y)$, où $X$ et $Y$ sont des v.a.r. gaussiennes indépendantes $\mathcal{N}(0,1)$. La v.a. $X^2 + Y^2 \sim \text{Exp}(1/2)$. L'algorithme de Box-Muller simule ce vecteur gaussien à partir de générateurs aléatoires uniformes $U$ et $V$.  Par changement de variables en polaire. Et sans utiliser la fonction de répartition $F$, ni son inverse $F^{-1}$ (qui ne sont pas explicites).

```python
N=10**6; #### D'abord, la répartition de deux échantillons gaussiens indépendants dans le plan, 
# par la commande "directe" normal().
XX=normal(0, 1, N);
YY=normal(0, 1, N);
plot(XX,YY,'.')
```

```python
hist(XX**2 + YY**2, bins = linspace(0,10,51), density = True);
xx=linspace(0,10,51); 
plot(xx, exp(-xx/2)/2 ,'r')
```

```python
U=rand(N); V=2*pi*rand(N);
X = sqrt(-2*log(U)) * cos(V) 
Y = sqrt(-2*log(U)) * sin(V)

figure(1)
hist(X, bins = linspace(-4,4,51), density = True,label='Histogramme de X');
xx=linspace(-4,4,121); 
plot(xx, exp(-xx**2/2)/(sqrt(2*pi)), 'r',label='densité de la gaussienne')
legend()
show()
figure(2)
hist(Y, bins = linspace(-4,4,51), density = True, label='Histogramme de Y');
xx=linspace(-4,4,121); 
plot(xx, exp(-xx**2/2)/(sqrt(2*pi)), 'r')
legend()
show()
```

```python
figure(3)
plot(X,Y,'.',label='repartition de l echantillon (X,Y) par Box-Muller')
legend()
show()
```

```python
#### Et pour aller plus loin : un histogramme 2-dimensionnel comparé au graphe de la densité sur R^2
### (Illustration hors programme pour le CCTP)

#from mpl_toolkits.mplot3d import axes3d

fig = figure()
ax1 = fig.add_subplot(121,projection='3d')
ax2 = fig.add_subplot(122,projection='3d')
Xmin=min(X)
Xmax=max(X)
Ymin=min(Y)
Ymax=max(Y)
hhist, xedges, yedges = histogram2d(X, Y, bins=20, range=[[Xmin, Xmax], [Ymin, Ymax]],density=True)
xpos, ypos = meshgrid(xedges[:-1], yedges[:-1], indexing="ij")
ZZ=(1/(2*pi))*exp(-(xpos**2+ypos**2)) #### xpos,ypos et ZZ sont "2-dimensionnels"
xposs=xpos.ravel()
yposs=ypos.ravel()
Ztop=hhist.ravel()
Zbottom=zeros_like(Ztop)
#### xposs,yposs et Ztop sont "1-dimensionnels"
dx=(xedges[1]-xedges[0])
dy=(yedges[1]-yedges[0])

ax1.bar3d(xposs, yposs, Zbottom, dx, dy, Ztop,shade=True) #### En bleu, le tracé de l'histogramme 2-dim

ax2.plot_surface(xpos,ypos,ZZ, edgecolor='r', lw=0.5,alpha=0.2) #### En rouge, le tracé du 
                                                                ### graphe de la densité sur R^2

show()
```
